# ScoreLens: a supplied prediction under a known synthetic failure
Run from the repository root in the installed local environment. This example creates synthetic inputs, runs the pinned T1 scorer, and inspects named discrepancies. It does not download data. The default uniformly samples at most 256 cells per file. These are local diagnostic measurements, not a hidden-target leaderboard score.
The generated directory is temporary; copy reports you want to keep to your own project folder.

```python
from pathlib import Path
import tempfile
from vec_scorelens.synthetic import write_case
from vec_scorelens.cli import run
work = Path(tempfile.mkdtemp(prefix='vec-scorelens-demo-'))
write_case(work / 'inputs', case='direction', seed=0)
print(work)
```

```python
report = run(work / 'inputs' / 'prediction.h5ad',
             work / 'inputs' / 'target.h5ad',
             work / 'inputs' / 'reference.h5ad',
             work / 'report', seed=0, max_cells=256)
print(report)
```

```python
import re
from IPython.display import Image, Markdown, display
summary = (report / 'summary.md').read_text(encoding='utf-8')
pending = []
for line in summary.splitlines(keepends=True):
    image = re.fullmatch(r'!\[.*\]\((figures/[^)]+\.png)\)\n?', line)
    if image:
        if pending:
            display(Markdown(''.join(pending)))
            pending = []
        display(Image(filename=str(report / image.group(1))))
    else:
        pending.append(line)
if pending:
    display(Markdown(''.join(pending)))
```

## Read the evidence type before interpreting a table
Mean log-expression differences are not conventional fold changes. Local significant DE sets and the scorer's target-sized ranked selections answer different questions. Pair discrepancies sample a fractional absolute difference; they are not gene-regulatory interactions. Probe-class proportions depend on the supplied target labels and the fitted classifier.
The MMD experiment changes at most five genes selected by mean discrepancy, then reruns the same metric. Its result is sensitivity to that artificial intervention, not causal ownership or a prescription for improving a model. Target-dependent diagnostic adjustments must not be mistaken for a legitimate prediction on unseen data.
An existing report directory is protected from overwrite: choose a fresh directory for each run. Use only local data you are permitted to analyze; keep reports on restricted data private.
